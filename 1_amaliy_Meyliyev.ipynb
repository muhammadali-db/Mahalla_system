````python
import pandas as pd
import random


ismlar = ["Ali", "Vali", "Sardor", "Jasur", "Nodir", "Laylo", "Madina", "Zilola", "Guli", "Shahnoza"]
familiyalar = ["Karimov", "Aliyev", "Toshpulatov", "Solihov", "Rahimov", "Hasanova", "Usmonova", "Sadirova", "Sultonova", "Abdukarimova"]

bemor_ismlari = []

genders = ["M", "F", "M", "F", "F", "M", "M", "F", "M", "F"]

for g in genders:
    if g == "M":
        ism = random.choice(["Ali Aliyev", "Vali Karimov", "Sardor Toshpulatov", "Jasur Rahimov", "Nodir Solihov"])
    else:
        ism = random.choice(["Laylo Hasanova", "Madina Usmonova", "Zilola Sadirova", "Guli Sultonova", "Shahnoza Abdukarimova"])
    bemor_ismlari.append(ism)

data = {
    "Bemor_ID": ["P001", "P002", "P003", "P004", "P005",
                   "P006", "P007", "P008", "P009", "P010"],
    "To\'liq_Ism": bemor_ismlari,
    "Yosh": [25, 31, 45, 52, 29, 38, 60, 41, 34, 47],
    "Jins": genders,
    "Temperatura": [38.2, 37.1, 36.7, 37.8, 36.9, 38.5, 37.0, 36.6, 37.4, 38.0],
    "Qon_Bosimi": ["120/80", "130/85", "125/82", "140/90", "118/76",
                       "135/88", "145/95", "122/80", "128/84", "132/86"]
}

df = pd.DataFrame(data)
df
````

# PANDAS kutubxonasi yordamida hisoblash

````python
ortacha_yosh = df["Yosh"].mean()
ortacha_temp = df["Temperatura"].mean()
eng_baland_temp = df["Temperatura"].max()

print("Bemorlarning o'rtacha yoshi:",ortacha_yosh)
print("O'rtacha harorat:", round(ortacha_temp, 2))
print("Maksimal harorat:", eng_baland_temp)

````

**bold text** 1. DataFrame'ni ba'zi yetishmayotgan (missing) qiymatlar bilan qayta yaratish
Ushbu qadamda keyingi tahlillar va tozalash amallarini ko'rish uchun jadvalga ataylab ba'zi bo'sh (`None`/`nan`) qiymatlar kiritamiz.

````python
import pandas as pd
import numpy as np


data_missing = {
    "Bemor_ID": ["P001", "P002", "P003", "P004", "P005", "P006", "P007", "P008", "P009", "P010"],
    "To'liq_Ism": ["Nodir Solihov", "Shahnoza Abdukarimova", "Nodir Solihov", "Guli Sultonova", "Madina Usmonova", "Sardor Toshpulatov", "Nodir Solihov", "Shahnoza Abdukarimova", "Vali Karimov", "Guli Sultonova"],
    "Yosh": [25, None, 45, 52, 29, None, 60, 41, 34, 47],  # Ba'zi yoshlar bo'sh
    "Jins": ["M", "F", "M", None, "F", "M", None, "F", "M", "F"],  # Ba'zi jinslar bo'sh
    "Temperatura": [38.2, 37.1, None, 37.8, 36.9, 38.5, 37.0, None, 37.4, 38.0],  # Ba'zi haroratlar bo'sh
    "Qon_Bosimi": ["120/80", "130/85", "125/82", "140/90", "118/76", "135/88", "145/95", "122/80", "128/84", "132/86"]
}

df_missing = pd.DataFrame(data_missing)
display(df_missing)
````

# 3 va 4. Yetishmayotgan qiymatlarni topish va ularning sonini aniqlash

````python
print("Yetishmayotgan qiymatlar jadvali (True - yetishmaydi, False - mavjud):")
display(df_missing.isnull())

print("\nHar bir ustundagi yetishmayotgan qiymatlar soni:")
print(df_missing.isnull().sum())
````

# 5 va 6. Yetishmayotgan qiymatlarni to'ldirish va yakuniy jadvalni chiqarish
* Yoshni — o'rtacha yosh bilan;
* Jinsni — "Unknown" qiymati bilan;
* Haroratni (Temperatura) — o'rtacha harorat bilan to'ldiramiz.
```

````python

df_clean = df_missing.copy()


ortacha_yosh_qiymat = df_clean["Yosh"].mean()
ortacha_temp_qiymat = df_clean["Temperatura"].mean()


df_clean["Yosh"] = df_clean["Yosh"].fillna(ortacha_yosh_qiymat)
df_clean["Jins"] = df_clean["Jins"].fillna("Unknown")
df_clean["Temperatura"] = df_clean["Temperatura"].fillna(ortacha_temp_qiymat)


print("Tozalangan yakuniy jadval:")
display(df_clean)
````
